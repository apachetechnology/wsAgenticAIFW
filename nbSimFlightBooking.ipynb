# Flight Booking Scam

## Scripted

```python
from sim_flight_booking.run_scenario2 import CFlightScamScenarioScripted

objFSSS = CFlightScamScenarioScripted()
baseline = objFSSS.run_baseline()
mitigated = objFSSS.run_mitigated()

print("\n" + "=" * 80)
print("COMPARISON")
print("=" * 80)
print(f"Baseline  booking result:  {baseline['booking_result']}")
print(f"Mitigated booking result:  {mitigated['booking_result']}")
```

## With LLM

```python
from api_server.Ollama_server import COllamaServer
from sim_flight_booking.run_scenario2 import CFlightScamScenarioLLM

srv = COllamaServer()
srv.start_server()      # starts "ollama serve" if not already running
srv.get_client()        # must be called, else srv.mClient does not exist

model = "llama3.2:1b"   # or "gemma3:1b"; run `ollama pull <model>` beforehand
seed = 0                # any int; vary 0..19 for repeated trials

objFSSS_LLM = CFlightScamScenarioLLM()
baseline = objFSSS_LLM.run_baseline(srv, model, seed)
mitigated = objFSSS_LLM.run_mitigated(srv, model, seed)

print("\n" + "=" * 80)
print("COMPARISON (for the paper's empirical note)")
print("=" * 80)
print(f"Baseline  booking result:  {baseline['booking_result']}")
print(f"Mitigated booking result:  {mitigated['booking_result']}")
```
